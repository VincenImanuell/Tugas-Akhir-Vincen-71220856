## Data Load From YFinance

```python
"""
Filter saham IDX aktif
Pakai download satu per satu dengan fallback minimal
"""

import yfinance as yf
import pandas as pd
import time
import random

sektor = terkelompok

# ─── DATE RANGES ──────────────────────────────────────────────────────────────
data_start  = "2021-06-01"
data_end    = "2026-01-01"
train_start = "2022-01-01"
train_end   = "2023-12-31"
test_start  = "2024-01-01"
test_end    = "2025-12-31"

STAGNANT_DAYS = 21   # ~1 bulan hari bursa (Batas maksimal harga boleh beku)
MIN_DATA      = 50   # Minimal riwayat data (baris)
STAGNANT_CAL  = 35   # ~1 bulan kalender (Toleransi gap dari 31 Des 2023)

all_tickers = [t for lst in sektor.values() for t in lst]
ticker_to_sektor = {t: s for s, lst in sektor.items() for t in lst}

print(f"Total ticker: {len(all_tickers)}")
print("Download satu per satu (filter stagnan 1 bulan)...\n")

aktif   = []
dibuang = []
errors  = []
hist_dict = {}

# PROSES FILTERING
for i, ticker in enumerate(all_tickers):
    sym = ticker + ".JK"
    try:
        # Download keseluruhan data 2020-2025
        hist = yf.download(sym, start="2020-01-01", end="2025-12-31",
                           multi_level_index=False, auto_adjust=False, progress=False)

        # Filtering hanya pakai data close sampai akhir 2023
        hist_filter = hist.loc[:"2023-12-31"]

        close = hist_filter["Close"].dropna() if not hist_filter.empty else pd.Series(dtype=float)

        stagnant = False
        alasan = ""

        if len(close) < MIN_DATA:
            stagnant = True
            alasan = f"data kurang ({len(close)} titik)"
        else:
            # Patokan di akhir tahun 2023
            tanggal_patokan = pd.Timestamp("2023-12-31")

            last_date = close.index[-1]
            if hasattr(last_date, 'tzinfo') and last_date.tzinfo is not None:
                last_date = last_date.tz_convert(None)

            gap = (tanggal_patokan - last_date).days

            if gap > STAGNANT_CAL:
                stagnant = True
                alasan = f"tidak ada transaksi {gap} hari (dari 31 Des 2023)"
            else:
                # Cek apakah pernah suspend/beku > 1 bulan di riwayatnya
                roll_std = close.rolling(STAGNANT_DAYS).std()
                if (roll_std == 0).any():
                    stagnant = True
                    alasan = "pernah suspend/harga beku > 1 bulan"

        if stagnant:
            dibuang.append((ticker, alasan))
        else:
            aktif.append(ticker)
            hist_dict[ticker] = hist

        status = "❌ dibuang" if stagnant else "✅ aktif"
        print(f"  [{i+1:>3}/{len(all_tickers)}] {ticker:<6} {status} {alasan}")

    except Exception as e:
        errors.append(ticker)
        print(f"  [{i+1:>3}/{len(all_tickers)}] {ticker:<6} ⚠️  error: {str(e)[:60]}")

    time.sleep(random.uniform(0.3, 0.7))

# ─── OUTPUT ────────────────────────────────────────────────────
aktif_set = set(aktif)
sektor_bersih = {s: [t for t in lst if t in aktif_set]
                 for s, lst in sektor.items()
                 if any(t in aktif_set for t in lst)}

print("\n" + "="*55)
print(f"  TOTAL SEMULA  : {len(all_tickers)}")
print(f"  AKTIF         : {len(aktif)}")
print(f"  DIBUANG       : {len(dibuang)}")
print(f"  ERROR         : {len(errors)}")
print("="*55)

for s, lst in sektor_bersih.items():
    print(f"  {s:<30} {len(lst):>3} / {len(sektor[s])}")
```

```python
import pandas as pd
from itertools import combinations

pearson_results = {}

for sektor, df_sektor in cleaned_df.items():

    # Filter hanya pakai data 2020-01-01 sampai 2023-12-31
    df_train = df_sektor.loc["2020-01-01":"2023-12-31"]

    # hitung matriks korelasi
    corr_matrix = df_train.corr(method="pearson")

    pairs = []

    # ambil hanya kombinasi unik (tanpa duplikasi & diagonal)
    for saham1, saham2 in combinations(corr_matrix.columns, 2):
        corr_value = corr_matrix.loc[saham1, saham2]
        pairs.append((saham1, saham2, corr_value))

    # ubah jadi DataFrame
    pair_df = pd.DataFrame(pairs, columns=["Saham_1", "Saham_2", "Correlation"])

    # urutkan dari tertinggi
    pair_df = pair_df.sort_values(by="Correlation", ascending=False)

    pearson_results[sektor] = pair_df

    print(sektor, "→ total pair:", len(pair_df))
```

## Cointegration

```python
import pandas as pd
from itertools import combinations
from statsmodels.tsa.stattools import coint

cointegration_results = {}
top5_per_sektor = {}

for sektor, df_sektor in cleaned_df.items():
    print(f"\n=== Sektor: {sektor} (Two-Way Screening) ===")

    # Filter data training
    df_train = df_sektor.loc["2020-01-01":"2023-12-31"]

    pairs = []
    unique_combinations = list(combinations(df_train.columns, 2))
    total_pairs = len(unique_combinations)
    counter = 0

    for saham1, saham2 in unique_combinations:
        counter += 1
        series1 = df_train[saham1]
        series2 = df_train[saham2]

        score12, pval12, _ = coint(series1, series2)
        score21, pval21, _ = coint(series2, series1)

        if pval12 < pval21:
            best_y, best_x, best_score, best_pval = saham1, saham2, score12, pval12
        else:
            best_y, best_x, best_score, best_pval = saham2, saham1, score21, pval21

        print(f"[{counter}/{total_pairs}] Best orientation: {best_y}-{best_x} | p-value: {best_pval:.12e}")

        pairs.append((best_y, best_x, best_score, best_pval))

    pair_df = pd.DataFrame(pairs, columns=["Saham_Y", "Saham_X", "Test_Statistic", "P_Value"])
    pair_df = pair_df.sort_values(by="P_Value").reset_index(drop=True)
    cointegration_results[sektor] = pair_df

    significant = pair_df[pair_df["P_Value"] < 0.05]
    top5 = significant.head(5).reset_index(drop=True)
    top5_per_sektor[sektor] = top5

    print(f"\n  → Top 5 cointegrated pairs ({sektor}):")
    if top5.empty:
        print("    Tidak ada pair cointegrated (p < 0.05)")
    else:
        print(top5[["Saham_Y", "Saham_X", "P_Value"]].to_string())
```

```python
list_top5 = [df.assign(Sektor=s) for s, df in top5_per_sektor.items() if not df.empty]

if list_top5:
    all_top5 = pd.concat(list_top5, ignore_index=True)
    all_top5 = all_top5[["Sektor", "Saham_Y", "Saham_X", "Test_Statistic", "P_Value"]]

    print("\n=== Rekap Semua Top 5 Per Sektor ===")
    print(all_top5.to_string(index=True))

    all_top5.to_csv("top5_pairs_per_sektor_twoway.csv", index=False)
    print("\nDisimpan ke top5_pairs_per_sektor_twoway.csv")
else:
    print("\nTidak ada pasangan yang signifikan secara statistik di seluruh sektor.")
```

# Pair Trading (dengan grid search)

```python
import os
import numpy as np
import pandas as pd
import statsmodels.api as sm
from statsmodels.regression.rolling import RollingOLS
from statsmodels.tsa.stattools import adfuller
import yfinance as yf
from itertools import product as iproduct
from openpyxl import Workbook
from openpyxl.styles import Font, PatternFill, Alignment, Border, Side
from openpyxl.utils import get_column_letter

os.makedirs("BT_RSI", exist_ok=True)

# ─── PAIR CLASSIFICATION ───────────────────────────────────────────────────────
COINTEGRATION_PAIRS = [
    "SOSS-SKRN", "KOPI-DWGL", "PICO-ALMI", "JAST-OASA",
    "AKSI-HELI", "JMAS-BBRI", "NZIA-PAMG", "ATIC-DIVA",
    "SKBM-MGRO", "GDYR-GLOB", "MIKA-PRIM",
]

PEARSON_SPEARMAN_PAIRS = [
    "HEXA-IMPC",
    "ITMG-ADRO",
    "SPMA-UNIC",
    "ADHI-PTPP",
    "SMDR-TMAS",
    "PNBN-PNLF", "BBNI-BMRI",
    "POLL-URBN",
    "MLPT-DMMX",
    "ADES-AMRT",
    "MAPI-PANR",
    "INAF-KAEF",
]

all_pairs        = COINTEGRATION_PAIRS + PEARSON_SPEARMAN_PAIRS
COINTEGRATION_SET = set(COINTEGRATION_PAIRS)

# ─── DATE RANGES ──────────────────────────────────────────────────────────────
data_start  = "2021-06-01"   # buffer untuk Rolling OLS
train_start = "2022-01-01"
train_end   = "2023-12-31"
test_start  = "2024-01-01"
test_end    = "2025-12-31"

# ─── GRID SEARCH PARAMETERS ───────────────────────────────────────────────────
entry_thresholds = [1.5, 2.0, 2.5]
exit_thresholds  = [0.0, 0.25, 0.5]
window_list      = [20, 40, 60]

sl_tp_pairs = [
    (0.02, 0.03), (0.02, 0.04),
    (0.03, 0.045), (0.03, 0.06),
    (0.04, 0.06), (0.04, 0.08),
    (0.05, 0.075), (0.05, 0.10),
    (0.03, 999999), (0.05, 999999), (0.07, 999999)
]

rsi_periods    = [7, 10, 14]
rsi_thresholds = [25, 30, 35, 40]

MIN_TRAIN_TRADES = 10
OLS_WINDOW       = 60

# ─── HELPER: Tentukan arah Y/X berdasarkan ADF p-value residual OLS ──────────
def determine_direction_by_adf(price_a, price_b, ticker_a, ticker_b):
    def ols_residual_adf(dep, indep):
        X_const  = sm.add_constant(indep)
        model    = sm.OLS(dep, X_const).fit()
        residual = model.resid
        try:
            adf_result = adfuller(residual.dropna(), autolag="AIC")
            return float(adf_result[1])
        except Exception:
            return 1.0

    pval_ab = ols_residual_adf(price_a, price_b)
    pval_ba = ols_residual_adf(price_b, price_a)

    if pval_ab <= pval_ba:
        return price_a, price_b, ticker_a, ticker_b, pval_ab, pval_ba, "AB"
    else:
        return price_b, price_a, ticker_b, ticker_a, pval_ba, pval_ab, "BA"

# ─── HELPERS ──────────────────────────────────────────────────────────────────
def fmt_sl_tp(v):
    return "Disabled" if v >= 999 else f"{v*100:.0f}%"

def calc_rsi(series, period):
    delta    = series.diff()
    gain     = delta.clip(lower=0)
    loss     = -delta.clip(upper=0)
    sum_gain = gain.rolling(window=period).sum()
    sum_loss = loss.rolling(window=period).sum()
    total    = (sum_gain + sum_loss).replace(0, np.nan)
    return (sum_gain / total) * 100

def calc_zscore(df_in, win):
    spread = df_in["Dynamic_Spread"]
    mean   = spread.rolling(win).mean()
    std    = spread.rolling(win).std()
    return spread, (spread - mean) / std

def run_backtest(df, z_series, z_a, z_b, z_ex, sl, tp,
                 rsi_y=None, rsi_x=None, rsi_threshold=None):
    use_rsi  = (rsi_y is not None) and (rsi_x is not None) and (rsi_threshold is not None)
    position = pd.Series(0, index=df.index, dtype=int)
    exit_rsn = pd.Series("", index=df.index, dtype=object)
    entry_pr = None
    active_p = 0

    for i in range(1, len(df)):
        idx = df.index[i]
        z   = z_series.iloc[i]
        py  = df["Y"].iloc[i]
        px  = df["X"].iloc[i]

        if pd.isna(z):
            continue

        exit_reason = ""
        closed      = False

        if active_p != 0 and entry_pr is not None:
            cur_price       = py if active_p == 1 else px
            gross_pnl       = (cur_price - entry_pr) / entry_pr
            transaction_fee = 0.004
            pnl             = gross_pnl - transaction_fee

            if pnl <= -sl:
                active_p = 0; entry_pr = None; exit_reason = "SL"; closed = True
            elif pnl >= tp:
                active_p = 0; entry_pr = None; exit_reason = "TP"; closed = True

            if not closed and abs(z) < z_ex:
                active_p = 0; entry_pr = None; exit_reason = "Z-Exit"; closed = True

            if not closed and i == len(df) - 1:
                active_p = 0; entry_pr = None; exit_reason = "End of Data"; closed = True

        if active_p == 0 and not closed and i < len(df) - 1:
            if z < z_b:
                rsi_ok = True
                if use_rsi:
                    ry     = rsi_y.iloc[i]
                    rsi_ok = (not pd.isna(ry)) and (ry <= rsi_threshold)
                if rsi_ok:
                    active_p = 1; entry_pr = py
            elif z > z_a:
                rsi_ok = True
                if use_rsi:
                    rx     = rsi_x.iloc[i]
                    rsi_ok = (not pd.isna(rx)) and (rx <= rsi_threshold)
                if rsi_ok:
                    active_p = -1; entry_pr = px

        position.iloc[i] = active_p
        exit_rsn.iloc[i] = exit_reason

    prev_pos  = position.shift(1).fillna(0)
    strat_ret = pd.Series(0.0, index=df.index)
    strat_ret[prev_pos == 1]  = df["Ret_Y"]
    strat_ret[prev_pos == -1] = df["Ret_X"]
    strat_ret = strat_ret.fillna(0)

    equity  = (1 + strat_ret).cumprod()
    mean_d  = strat_ret.mean()
    std_d   = strat_ret.std()

    return {
        "position":   position,
        "exit_rsn":   exit_rsn,
        "strat_ret":  strat_ret,
        "equity":     equity,
        "sharpe":     0 if std_d == 0 else (mean_d / std_d) * np.sqrt(252),
        "total_ret":  equity.iloc[-1] - 1,
        "mdd":        (equity / equity.cummax() - 1).min(),
        "win_rate":   (strat_ret > 0).sum() / max(1, (strat_ret != 0).sum()),
        "volatility": std_d * np.sqrt(252),
        "n_trades":   int(((position != 0) & (position != position.shift(1))).sum()),
        "n_sl":       (exit_rsn == "SL").sum(),
        "n_tp":       (exit_rsn == "TP").sum(),
        "n_ze":       (exit_rsn == "Z-Exit").sum(),
    }

def build_trade_log(df, res):
    trade_log  = []
    in_trade   = False
    entry_date = entry_pos = entry_y = entry_x = None

    for i in range(1, len(df)):
        prev = res["position"].iloc[i-1]
        curr = res["position"].iloc[i]
        if prev == 0 and curr != 0:
            in_trade   = True
            entry_date = df.index[i]
            entry_pos  = curr
            entry_y    = float(df["Y"].iloc[i])
            entry_x    = float(df["X"].iloc[i])
        elif prev != 0 and curr == 0 and in_trade:
            exit_date = df.index[i]
            exit_y    = float(df["Y"].iloc[i])
            exit_x    = float(df["X"].iloc[i])
            gross_pnl = (exit_y - entry_y)/entry_y if entry_pos == 1 else (exit_x - entry_x)/entry_x
            pnl       = gross_pnl - 0.004
            exit_r    = res["exit_rsn"].iloc[i]
            trade_log.append([
                entry_date.date(), exit_date.date(),
                "Long Y" if entry_pos == 1 else "Long X",
                round(entry_y,4), round(exit_y,4),
                round(entry_x,4), round(exit_x,4),
                f"{pnl*100:.2f}%", exit_r,
                (exit_date - entry_date).days
            ])
            in_trade = False

    if in_trade and entry_date is not None:
        exit_date = df.index[-1]
        exit_y    = float(df["Y"].iloc[-1])
        exit_x    = float(df["X"].iloc[-1])
        gross_pnl = (exit_y - entry_y)/entry_y if entry_pos == 1 else (exit_x - entry_x)/entry_x
        pnl       = gross_pnl - 0.004
        trade_log.append([
            entry_date.date(), exit_date.date(),
            "Long Y" if entry_pos == 1 else "Long X",
            round(entry_y,4), round(exit_y,4),
            round(entry_x,4), round(exit_x,4),
            f"{pnl*100:.2f}%", "End of Data",
            (exit_date - entry_date).days
        ])
    return trade_log

def make_styles():
    return {
        "header_font":    Font(name="Arial", bold=True, color="FFFFFF"),
        "header_fill":    PatternFill("solid", start_color="1F4E79"),
        "center_align":   Alignment(horizontal="center"),
        "left_align":     Alignment(horizontal="left"),
        "thin_border":    Border(left=Side(style="thin"), right=Side(style="thin"),
                                 top=Side(style="thin"),  bottom=Side(style="thin")),
        "green_fill":     PatternFill("solid", start_color="E2EFDA"),
        "red_fill":       PatternFill("solid", start_color="FFDAD9"),
        "title_fill":     PatternFill("solid", start_color="1F4E79"),
        "section_fill":   PatternFill("solid", start_color="2E75B6"),
        "label_fill":     PatternFill("solid", start_color="D6E4F0"),
        "gold_fill":      PatternFill("solid", start_color="FFF2CC"),
        "gs_fill":        PatternFill("solid", start_color="375623"),
        "buy_y_fill":     PatternFill("solid", start_color="DDEEFF"),
        "buy_x_fill":     PatternFill("solid", start_color="E8F5E9"),
        "exit_fill":      PatternFill("solid", start_color="FFF9C4"),
        "hold_fill":      PatternFill("solid", start_color="FFFFFF"),
        "rsi_sect_fill":  PatternFill("solid", start_color="7030A0"),
        "rsi_label_fill": PatternFill("solid", start_color="EAD8F5"),
        "better_fill":    PatternFill("solid", start_color="C6EFCE"),
        "worse_fill":     PatternFill("solid", start_color="FFDAD9"),
    }

def style_header_row(ws, s):
    for cell in ws[1]:
        cell.font      = s["header_font"]
        cell.fill      = s["header_fill"]
        cell.alignment = s["center_align"]
        cell.border    = s["thin_border"]

def auto_col_width(ws, extra=4):
    for col in ws.columns:
        width = max(len(str(c.value or "")) for c in col) + extra
        ws.column_dimensions[get_column_letter(col[0].column)].width = width

def write_excel(wb, A, B, df, res_base, res_rsi, tlog_base, tlog_rsi, gs_base_df, gs_rsi_df,
                window, z_atas, z_exit, sl_best, tp_best, rsi_period_best, rsi_thresh_best,
                rsi_y_series=None, rsi_x_series=None, tlog_train_base=None, tlog_train_rsi=None,
                pair_type="cointegration", direction_info=None):

    s    = make_styles()
    bh_y = df["BH_Y"].iloc[-1] - 1
    bh_x = df["BH_X"].iloc[-1] - 1
    test_start_str = df.index[0].strftime('%Y-%m-%d')
    test_end_str   = df.index[-1].strftime('%Y-%m-%d')

    ws1 = wb.active
    ws1.title = "Comparison Summary"

    direction_label = "Cointegration (pre-set)"
    if pair_type == "pearson_spearman" and direction_info is not None:
        pval_chosen = direction_info["pval_chosen"]
        pval_alt    = direction_info["pval_alt"]
        direction_label = (f"Pearson/Spearman → ADF: {A}~{B}={pval_chosen:.4f} vs "
                           f"{B}~{A}={pval_alt:.4f} → pilih {direction_info['dir_code']}")

    rows = [
        ["PAIR TRADING — STRATEGY EVALUATION (TEST DATA)", "", "", ""],
        ["", "", "", ""],
        ["Metric", "Base Strategy", "RSI-Filtered", "Δ (RSI − Base)"],
        ["Total Return (%)", f"{res_base['total_ret']*100:.2f}%", f"{res_rsi['total_ret']*100:.2f}%", f"{(res_rsi['total_ret']-res_base['total_ret'])*100:.2f}%"],
        ["Sharpe Ratio", f"{res_base['sharpe']:.4f}", f"{res_rsi['sharpe']:.4f}", f"{res_rsi['sharpe']-res_base['sharpe']:.4f}"],
        ["Max Drawdown (%)", f"{res_base['mdd']*100:.2f}%", f"{res_rsi['mdd']*100:.2f}%", f"{(res_rsi['mdd']-res_base['mdd'])*100:.2f}%"],
        ["Win Rate (%)", f"{res_base['win_rate']*100:.2f}%", f"{res_rsi['win_rate']*100:.2f}%", f"{(res_rsi['win_rate']-res_base['win_rate'])*100:.2f}%"],
        ["Annual Volatility (%)", f"{res_base['volatility']*100:.2f}%", f"{res_rsi['volatility']*100:.2f}%", f"{(res_rsi['volatility']-res_base['volatility'])*100:.2f}%"],
        ["Total Trades", len(tlog_base), len(tlog_rsi), len(tlog_rsi) - len(tlog_base)],
        ["Avg Hold Days", f"{sum(t[9] for t in tlog_base)/max(1,len(tlog_base)):.1f}", f"{sum(t[9] for t in tlog_rsi)/max(1,len(tlog_rsi)):.1f}", ""],
        ["Best Trade", f"{max((float(t[7].replace('%','')) for t in tlog_base),default=0):.2f}%", f"{max((float(t[7].replace('%','')) for t in tlog_rsi),default=0):.2f}%", ""],
        ["Worst Trade", f"{min((float(t[7].replace('%','')) for t in tlog_base),default=0):.2f}%", f"{min((float(t[7].replace('%','')) for t in tlog_rsi),default=0):.2f}%", ""],
        ["Avg Trade Return", f"{sum(float(t[7].replace('%','')) for t in tlog_base)/max(1,len(tlog_base)):.2f}%", f"{sum(float(t[7].replace('%','')) for t in tlog_rsi)/max(1,len(tlog_rsi)):.2f}%", ""],
        ["TP Hits", res_base['n_tp'], res_rsi['n_tp'], res_rsi['n_tp'] - res_base['n_tp']],
        ["SL Hits", res_base['n_sl'], res_rsi['n_sl'], res_rsi['n_sl'] - res_base['n_sl']],
        ["Z-Exits", res_base['n_ze'], res_rsi['n_ze'], res_rsi['n_ze'] - res_base['n_ze']],
        ["", "", "", ""],
        ["BENCHMARK (TEST DATA)", "", "", ""],
        [f"Buy & Hold Y ({A})", f"{bh_y*100:.2f}%", "", ""],
        [f"Buy & Hold X ({B})", f"{bh_x*100:.2f}%", "", ""],
        ["vs B&H Y (Base)",  f"{(res_base['total_ret']-bh_y)*100:.2f}%", "", ""],
        ["vs B&H Y (RSI)",   f"{(res_rsi['total_ret']-bh_y)*100:.2f}%",  "", ""],
        ["", "", "", ""],
        ["PARAMETERS USED (Optimized on Train)", "", "", ""],
        ["Stock Y (Dependen)", A, "", ""],
        ["Stock X (Independen)", B, "", ""],
        ["Pair Type", pair_type.replace("_", "/").title(), "", ""],
        ["Direction Selection", direction_label, "", ""],
        ["Train Period", f"{train_start} → {train_end}", "", ""],
        ["Test Period (OOS)", f"{test_start_str} → {test_end_str}", "", ""],
        ["z_entry (±)", z_atas, "", ""],
        ["z_exit", z_exit, "", ""],
        ["Z-Score Window", window, "", ""],
        ["Stop Loss (best)", fmt_sl_tp(sl_best), "", ""],
        ["Take Profit (best)", fmt_sl_tp(tp_best), "", ""],
        ["Transaction Fee", "0.40% per Round-Trip", "", ""],
        ["Hedge Ratio Model", "Dynamic (Rolling OLS, win=60)", "", ""],
        ["RR Ratio", f"1 : {tp_best/sl_best:.1f}" if sl_best < 999 and tp_best < 999 else "N/A", "", ""],
        ["RSI Period (best)", rsi_period_best, "", ""],
        ["RSI Threshold (best)", rsi_thresh_best, "", ""],
    ]

    for row_data in rows:
        ws1.append(row_data)

    metric_rows_delta = {
        "Total Return (%)": "higher_better", "Sharpe Ratio": "higher_better",
        "Max Drawdown (%)": "lower_better", "Win Rate (%)": "higher_better",
        "Annual Volatility (%)": "lower_better", "Total Trades": "neutral",
        "TP Hits": "higher_better", "SL Hits": "lower_better", "Z-Exits": "neutral",
    }

    label_rows = [
        "Avg Hold Days","Best Trade","Worst Trade","Avg Trade Return",
        f"Buy & Hold Y ({A})", f"Buy & Hold X ({B})",
        "vs B&H Y (Base)", "vs B&H Y (RSI)",
        "Stock Y (Dependen)", "Stock X (Independen)",
        "Pair Type", "Direction Selection",
        "Train Period", "Test Period (OOS)",
        "z_entry (±)", "z_exit", "Z-Score Window",
        "Stop Loss (best)", "Take Profit (best)", "Transaction Fee",
        "Hedge Ratio Model", "RR Ratio", "RSI Period (best)", "RSI Threshold (best)"
    ]

    for row in ws1.iter_rows(min_row=1, max_row=ws1.max_row):
        label = str(row[0].value or "")
        for cell in row:
            cell.border = s["thin_border"]; cell.alignment = s["left_align"]; cell.font = Font(name="Arial", size=11)

        if label == "PAIR TRADING — STRATEGY EVALUATION (TEST DATA)":
            for cell in row:
                cell.font = Font(name="Arial", bold=True, size=13, color="FFFFFF")
                cell.fill = s["title_fill"]
            ws1.merge_cells(f"A{row[0].row}:D{row[0].row}")
        elif label == "Metric":
            for cell in row:
                cell.font = Font(name="Arial", bold=True, color="FFFFFF")
                cell.fill = s["header_fill"]
                cell.alignment = s["center_align"]
        elif label in ["BENCHMARK (TEST DATA)", "PARAMETERS USED (Optimized on Train)"]:
            for cell in row:
                cell.font = Font(name="Arial", bold=True, color="FFFFFF")
                cell.fill = s["rsi_sect_fill"]
            ws1.merge_cells(f"A{row[0].row}:D{row[0].row}")
        elif label in metric_rows_delta:
            row[0].fill = s["label_fill"]
            row[0].font = Font(name="Arial", bold=True, size=11)
            direction = metric_rows_delta[label]
            if direction != "neutral":
                try:
                    dv = float(str(row[3].value).replace("%",""))
                    if dv != 0:
                        better = (direction == "higher_better" and dv > 0) or (direction == "lower_better" and dv < 0)
                        row[3].fill = s["better_fill"] if better else s["worse_fill"]
                except: pass
        elif label in label_rows:
            row[0].fill = s["label_fill"]
            row[0].font = Font(name="Arial", bold=True, size=11)

    for i, w in enumerate([38, 20, 20, 18], 1):
        ws1.column_dimensions[get_column_letter(i)].width = w

    tlog_headers = ["Entry Date","Exit Date","Direction","Entry Y","Exit Y","Entry X","Exit X","Net PnL","Exit Reason","Hold Days"]
    log_sheets   = [("Train Log (Base)", tlog_train_base), ("Train Log (RSI)", tlog_train_rsi),
                    ("Test Log (Base)", tlog_base), ("Test Log (RSI)", tlog_rsi)]

    for sheet_name, tlog in log_sheets:
        if tlog is None: continue
        wst = wb.create_sheet(sheet_name)
        wst.append(tlog_headers)
        style_header_row(wst, s)
        for row_data in tlog:
            wst.append(row_data)
            try:
                fill = s["green_fill"] if float(str(wst.cell(wst.max_row,8).value).replace("%","")) >= 0 else s["red_fill"]
            except: fill = s["green_fill"]
            for cell in wst[wst.max_row]:
                cell.fill = fill; cell.border = s["thin_border"]; cell.alignment = s["center_align"]
        auto_col_width(wst); wst.freeze_panes = "A2"

    ws4 = wb.create_sheet("Grid Search Train Z+SLTP")
    ws4.append(["Rank","z_entry","z_exit","Window","SL","TP","RR Ratio","Total Return (%)","Sharpe Ratio","Max Drawdown (%)","Win Rate (%)","Trades","SL Hits","TP Hits","Z-Exits"])
    style_header_row(ws4, s)
    for rank, rd in gs_base_df.reset_index().iterrows():
        sl_v = rd["SL"]; tp_v = rd["TP"]
        rr_str = (f"1 : {tp_v/sl_v:.1f}" if sl_v < 999 and tp_v < 999
                  else "No SL / No TP" if sl_v >= 999 and tp_v >= 999
                  else f"No SL / TP {fmt_sl_tp(tp_v)}" if sl_v >= 999
                  else f"SL {fmt_sl_tp(sl_v)} / No TP")
        row = [rank+1, rd["z_entry"], rd["z_exit"], int(rd["window"]),
               fmt_sl_tp(sl_v), fmt_sl_tp(tp_v), rr_str,
               rd["Total Return (%)"], rd["Sharpe Ratio"], rd["Max Drawdown (%)"],
               rd["Win Rate (%)"], rd["Trades"], rd["SL Hits"], rd["TP Hits"], rd["Z-Exits"]]
        ws4.append(row)
        fill = s["gold_fill"] if rank == 0 else s["hold_fill"]
        for cell in ws4[ws4.max_row]:
            cell.fill = fill; cell.font = Font(name="Arial", bold=(rank==0), size=11)
            cell.border = s["thin_border"]; cell.alignment = s["center_align"]
    auto_col_width(ws4); ws4.freeze_panes = "A2"

    ws5 = wb.create_sheet("Grid Search Train RSI")
    ws5.append(["Rank","RSI Period","RSI Threshold","Total Return (%)","Sharpe Ratio","Max Drawdown (%)","Win Rate (%)","Trades","SL Hits","TP Hits","Z-Exits"])
    style_header_row(ws5, s)
    for rank, rd in gs_rsi_df.reset_index().iterrows():
        row = [rank+1, int(rd["rsi_period"]), int(rd["rsi_threshold"]),
               rd["Total Return (%)"], rd["Sharpe Ratio"], rd["Max Drawdown (%)"],
               rd["Win Rate (%)"], rd["Trades"], rd["SL Hits"], rd["TP Hits"], rd["Z-Exits"]]
        ws5.append(row)
        fill = s["gold_fill"] if rank == 0 else s["hold_fill"]
        for cell in ws5[ws5.max_row]:
            cell.fill = fill; cell.font = Font(name="Arial", bold=(rank==0), size=11)
            cell.border = s["thin_border"]; cell.alignment = s["center_align"]
    auto_col_width(ws5); ws5.freeze_panes = "A2"

    ws6 = wb.create_sheet("Equity Curve")
    ws6.append(["Date","Base Strategy","RSI Strategy","Buy & Hold Y","Buy & Hold X"])
    style_header_row(ws6, s)
    for idx, row in df[["Equity_Base","Equity_RSI","BH_Y","BH_X"]].dropna().iterrows():
        ws6.append([idx.date(), round(float(row["Equity_Base"]),6), round(float(row["Equity_RSI"]),6),
                    round(float(row["BH_Y"]),6), round(float(row["BH_X"]),6)])
    for col in ws6.columns:
        ws6.column_dimensions[get_column_letter(col[0].column)].width = 18
    ws6.freeze_panes = "A2"

    sfm = {"Buy Y": s["buy_y_fill"], "Buy X": s["buy_x_fill"], "Exit": s["exit_fill"], "Hold": s["hold_fill"]}
    for sheet_name, res, include_rsi in [("Price & Signal (Base)", res_base, False), ("Price & Signal (RSI)", res_rsi, True)]:
        wsx    = wb.create_sheet(sheet_name)
        pos_s  = res["position"]
        prev_p = pos_s.shift(1).fillna(0)
        sig_s  = pd.Series("Hold", index=df.index)
        sig_s[(prev_p == 0) & (pos_s ==  1)] = "Buy Y"
        sig_s[(prev_p == 0) & (pos_s == -1)] = "Buy X"
        sig_s[(prev_p != 0) & (pos_s ==  0)] = "Exit"
        cum_pnl = (res["equity"] - 1) * 100
        strat_r = res["strat_ret"] * 100

        header_base = ["Date", f"Price Y ({A})", f"Price X ({B})", f"Z-Score (win={window})"]
        if include_rsi:
            header_base += [f"RSI Y (p={rsi_period_best})", f"RSI X (p={rsi_period_best})"]
        header_base += ["Signal", "Position", "Exit Reason", "Daily PnL (%)", "Cum PnL (%)"]
        wsx.append(header_base)
        style_header_row(wsx, s)

        for idx in df.index:
            row    = df.loc[idx]
            z_val  = row["Z"]
            sig    = sig_s.loc[idx]
            pos_lb = "Long Y" if int(pos_s.loc[idx]) == 1 else ("Long X" if int(pos_s.loc[idx]) == -1 else "-")
            ex_r   = res["exit_rsn"].loc[idx] if res["exit_rsn"].loc[idx] != "" else "-"

            data_row = [idx.date(), round(float(row["Y"]),4), round(float(row["X"]),4),
                        round(float(z_val),4) if pd.notna(z_val) else "-"]
            if include_rsi and rsi_y_series is not None and rsi_x_series is not None:
                ry = rsi_y_series.loc[idx] if idx in rsi_y_series.index else np.nan
                rx = rsi_x_series.loc[idx] if idx in rsi_x_series.index else np.nan
                ry_val = round(float(ry), 2) if pd.notna(ry) else "-"
                rx_val = round(float(rx), 2) if pd.notna(rx) else "-"
                data_row += [ry_val, rx_val]
            data_row += [sig, pos_lb, ex_r, round(float(strat_r.loc[idx]),4), round(float(cum_pnl.loc[idx]),4)]
            wsx.append(data_row)

            fill = sfm.get(sig, s["hold_fill"])
            for cell in wsx[wsx.max_row]:
                cell.fill = fill; cell.border = s["thin_border"]; cell.alignment = s["center_align"]

            if include_rsi and rsi_y_series is not None:
                for col_idx, rv in [(5, ry_val), (6, rx_val)]:
                    rc = wsx.cell(wsx.max_row, col_idx)
                    if isinstance(rv, float):
                        rc.fill = PatternFill("solid", start_color="FFD966") if rv <= rsi_thresh_best else s["hold_fill"]

        for i, w in enumerate([14,14,14,22,14,14,10,10,14,16,18] if include_rsi else [14,14,14,22,10,10,14,16,18], 1):
            wsx.column_dimensions[get_column_letter(i)].width = w
        wsx.freeze_panes = "A2"

# ── MAIN LOOP ──────────────────────────────────────────────────────────────────
failed_pairs  = []
direction_log = []

valid_z_combos    = [(ze, zx) for ze, zx in iproduct(entry_thresholds, exit_thresholds) if zx < ze]
total_base_combos = len(valid_z_combos) * len(window_list) * len(sl_tp_pairs)
total_rsi_combos  = len(rsi_periods) * len(rsi_thresholds)

print(f"Total pairs        : {len(all_pairs)}")
print(f"  Cointegration    : {len(COINTEGRATION_PAIRS)}")
print(f"  Pearson/Spearman : {len(PEARSON_SPEARMAN_PAIRS)}")
print(f"Train period       : {train_start} → {train_end}")
print(f"Test period (OOS)  : {test_start} → {test_end}")
print(f"Total base combos  : {total_base_combos}")
print(f"Total RSI combos   : {total_rsi_combos}")
print(f"Min trades (train) : {MIN_TRAIN_TRADES}")

for pair_idx, pair in enumerate(all_pairs):
    raw_A, raw_B     = pair.split("-")
    is_cointegration = pair in COINTEGRATION_SET

    print(f"\n{'='*65}")
    print(f"[{pair_idx+1}/{len(all_pairs)}] Processing: {raw_A} - {raw_B} "
          f"({'Cointegration' if is_cointegration else 'Pearson/Spearman'})")
    print(f"{'='*65}")

    try:
        # ── AMBIL DATA DARI all_df (per-ticker individual) ─────────────────────
        close_a = close_b = None
        for sektor_df in all_df.values():
            if raw_A in sektor_df.columns and close_a is None:
                close_a = sektor_df[raw_A].dropna()
            if raw_B in sektor_df.columns and close_b is None:
                close_b = sektor_df[raw_B].dropna()

        if close_a is None or close_b is None:
            missing = [t for t, s in [(raw_A, close_a), (raw_B, close_b)] if s is None]
            print(f"  Skip: ticker {missing} tidak ditemukan")
            failed_pairs.append((pair, f"ticker tidak ditemukan: {missing}"))
            continue

        # Filter rentang tanggal sejak buffer
        close_a = close_a.loc[data_start:data_end]
        close_b = close_b.loc[data_start:data_end]

        # Fix manual TMAS
        if raw_B == 'TMAS' or raw_A == 'TMAS':
            tmas_series = close_b if raw_B == 'TMAS' else close_a
            tmas_fix_date = pd.Timestamp("2023-05-23")
            if tmas_fix_date in tmas_series.index:
                tmas_series.loc[tmas_fix_date] = 294

        # Align hanya untuk pair ini
        common_idx = close_a.index.intersection(close_b.index)
        close_a    = close_a.loc[common_idx]
        close_b    = close_b.loc[common_idx]

        # ── TENTUKAN ARAH Y / X ────────────────────────────────────────────────
        direction_info = None

        if is_cointegration:
            A, B = raw_A, raw_B
            print(f"  [Cointegration] Arah tetap: Y={A}, X={B}")
        else:
            print(f"  [Pearson/Spearman] Menentukan arah terbaik via ADF p-value...")

            # Gunakan hanya data train untuk tentukan arah (konsisten dengan periode analisis)
            common_train = common_idx[(common_idx >= train_start) & (common_idx <= train_end)]
            close_a_train = close_a.loc[common_train]
            close_b_train = close_b.loc[common_train]

            (price_y, price_x, ticker_y, ticker_x,
             pval_chosen, pval_alt, direction_chosen) = determine_direction_by_adf(
                close_a_train, close_b_train, raw_A, raw_B
            )
            A, B = ticker_y, ticker_x

            direction_info = {
                "direction":   f"{A}~{B} (ADF p={pval_chosen:.4f}) vs {raw_B if A==raw_A else raw_A}~{A if A==raw_B else raw_B} (ADF p={pval_alt:.4f})",
                "pval_chosen": pval_chosen,
                "pval_alt":    pval_alt,
                "dir_code":    direction_chosen,
            }

            direction_log.append({
                "Pair (Original)":  pair,
                "Y (Dependen)":     A,
                "X (Independen)":   B,
                "ADF p-val Y~X":    round(pval_chosen, 6),
                "ADF p-val X~Y":    round(pval_alt, 6),
                "Direction Chosen": direction_chosen,
            })

            print(f"  ADF p-value  {raw_A}~{raw_B} = {pval_chosen:.6f}  |  {raw_B}~{raw_A} = {pval_alt:.6f}")
            print(f"  → Pilih: Y={A}, X={B}  (p-value terkecil = {min(pval_chosen, pval_alt):.6f})")

        # ── BUAT df_full ───────────────────────────────────────────────────────
        col_y = close_a if A == raw_A else close_b
        col_x = close_b if B == raw_B else close_a

        df_full = pd.concat([col_y.rename("Y"), col_x.rename("X")], axis=1).dropna()

        # ── ROLLING OLS (hedge ratio dinamis) ─────────────────────────────────
        X_full     = sm.add_constant(df_full["X"])
        roll_model = RollingOLS(df_full["Y"], X_full, window=OLS_WINDOW).fit()
        alpha_t    = roll_model.params["const"]
        beta_t     = roll_model.params["X"]
        df_full["Dynamic_Spread"] = df_full["Y"] - (alpha_t + beta_t * df_full["X"])

        # ── TRAIN / TEST SPLIT (sesuai date ranges) ────────────────────────────
        df_train = df_full.loc[train_start:train_end].copy()
        df_test  = df_full.loc[test_start:test_end].copy()

        if len(df_train) < 100 or len(df_test) < 50:
            print(f"  Skip: train={len(df_train)} rows, test={len(df_test)} rows")
            failed_pairs.append((pair, "train/test too short"))
            continue

        for dff in [df_train, df_test]:
            dff["Ret_Y"] = dff["Y"].pct_change().fillna(0)
            dff["Ret_X"] = dff["X"].pct_change().fillna(0)
            dff["BH_Y"]  = (1 + dff["Ret_Y"]).cumprod()
            dff["BH_X"]  = (1 + dff["Ret_X"]).cumprod()

        print(f"  Y={A} | X={B}")
        print(f"  Train : {df_train.index[0].date()} → {df_train.index[-1].date()} ({len(df_train)} rows)")
        print(f"  Test  : {df_test.index[0].date()} → {df_test.index[-1].date()} ({len(df_test)} rows)")

        # ── PHASE 1: Grid Search Z + SL/TP on TRAIN ───────────────────────────
        print(f"\n  [PHASE 1 - TRAIN] Grid search Z + SL/TP ({total_base_combos} combos)...")
        gs_base = []
        counter = 0

        for z_entry, z_ex, win, (sl, tp) in iproduct(entry_thresholds, exit_thresholds, window_list, sl_tp_pairs):
            if z_ex >= z_entry: continue
            counter += 1
            _, z_s_train = calc_zscore(df_train, win)
            r = run_backtest(df_train, z_s_train, z_entry, -z_entry, z_ex, sl, tp)
            gs_base.append({
                "z_entry": z_entry, "z_exit": z_ex, "window": win, "SL": sl, "TP": tp,
                "Total Return (%)": round(r["total_ret"]*100, 2),
                "Sharpe Ratio":     round(r["sharpe"], 4),
                "Max Drawdown (%)": round(r["mdd"]*100, 2),
                "Win Rate (%)":     round(r["win_rate"]*100, 2),
                "Trades": r["n_trades"], "SL Hits": r["n_sl"], "TP Hits": r["n_tp"], "Z-Exits": r["n_ze"],
            })
            if counter % 100 == 0 or counter == total_base_combos:
                print(f"    [{counter}/{total_base_combos}] ...", end="\r")

        gs_base_df  = pd.DataFrame(gs_base)
        valid_gs_df = gs_base_df[gs_base_df["Trades"] >= MIN_TRAIN_TRADES]
        if valid_gs_df.empty: valid_gs_df = gs_base_df[gs_base_df["Trades"] >= 5]
        if valid_gs_df.empty: valid_gs_df = gs_base_df

        gs_base_df   = valid_gs_df.sort_values("Sharpe Ratio", ascending=False).reset_index(drop=True)
        gs_base_df.index += 1

        best    = gs_base_df.iloc[0]
        z_atas  = best["z_entry"]; z_bawah = -best["z_entry"]; z_exit = best["z_exit"]
        window  = int(best["window"]); sl_best = best["SL"]; tp_best = best["TP"]
        print(f"\n  Best Z+SL/TP: z=±{z_atas} exit={z_exit} win={window} "
              f"SL={fmt_sl_tp(sl_best)} TP={fmt_sl_tp(tp_best)} Sharpe={best['Sharpe Ratio']:.4f} | Trades={int(best['Trades'])}")

        # ── PHASE 2: Grid Search RSI on TRAIN ─────────────────────────────────
        print(f"\n  [PHASE 2 - TRAIN] Grid search RSI ({total_rsi_combos} combos)...")
        gs_rsi  = []
        counter = 0
        _, z_best_train = calc_zscore(df_train, window)

        for rp, rt in iproduct(rsi_periods, rsi_thresholds):
            counter += 1
            rsi_y_train = calc_rsi(df_train["Y"], rp)
            rsi_x_train = calc_rsi(df_train["X"], rp)
            r = run_backtest(df_train, z_best_train, z_atas, z_bawah, z_exit, sl_best, tp_best,
                             rsi_y=rsi_y_train, rsi_x=rsi_x_train, rsi_threshold=rt)
            gs_rsi.append({
                "rsi_period": rp, "rsi_threshold": rt,
                "Total Return (%)": round(r["total_ret"]*100, 2),
                "Sharpe Ratio":     round(r["sharpe"], 4),
                "Max Drawdown (%)": round(r["mdd"]*100, 2),
                "Win Rate (%)":     round(r["win_rate"]*100, 2),
                "Trades": r["n_trades"], "SL Hits": r["n_sl"], "TP Hits": r["n_tp"], "Z-Exits": r["n_ze"],
            })
            if counter % 20 == 0: print(f"    [{counter}/{total_rsi_combos}] ...", end="\r")

        gs_rsi_df       = pd.DataFrame(gs_rsi).sort_values("Sharpe Ratio", ascending=False).reset_index(drop=True)
        gs_rsi_df.index += 1
        best_rsi        = gs_rsi_df.iloc[0]
        rsi_period_best = int(best_rsi["rsi_period"]); rsi_thresh_best = int(best_rsi["rsi_threshold"])
        print(f"\n  Best RSI: period={rsi_period_best}, threshold={rsi_thresh_best}, "
              f"Sharpe={best_rsi['Sharpe Ratio']:.4f} | Trades={int(best_rsi['Trades'])}")

        # ── PHASE 3: OUT-OF-SAMPLE TEST ───────────────────────────────────────
        print(f"\n  [PHASE 3 - TEST OOS] Applying frozen params to {test_start} → {test_end}...")
        _, z_best_test = calc_zscore(df_test, window)
        df_test["Z"]   = z_best_test

        res_test_base  = run_backtest(df_test, z_best_test, z_atas, z_bawah, z_exit, sl_best, tp_best)
        df_test["Equity_Base"] = res_test_base["equity"]
        tlog_test_base = build_trade_log(df_test, res_test_base)

        rsi_y_test_final = calc_rsi(df_test["Y"], rsi_period_best)
        rsi_x_test_final = calc_rsi(df_test["X"], rsi_period_best)
        res_test_rsi     = run_backtest(df_test, z_best_test, z_atas, z_bawah, z_exit, sl_best, tp_best,
                                        rsi_y=rsi_y_test_final, rsi_x=rsi_x_test_final, rsi_threshold=rsi_thresh_best)
        df_test["Equity_RSI"] = res_test_rsi["equity"]
        tlog_test_rsi    = build_trade_log(df_test, res_test_rsi)
        print(f"  Test Base: {len(tlog_test_base)} trades | Test RSI: {len(tlog_test_rsi)} trades")

        # ── TRAIN LOGS ─────────────────────────────────────────────────────────
        res_train_base  = run_backtest(df_train, z_best_train, z_atas, z_bawah, z_exit, sl_best, tp_best)
        tlog_train_base = build_trade_log(df_train, res_train_base)

        rsi_y_train_best = calc_rsi(df_train["Y"], rsi_period_best)
        rsi_x_train_best = calc_rsi(df_train["X"], rsi_period_best)
        res_train_rsi    = run_backtest(df_train, z_best_train, z_atas, z_bawah, z_exit, sl_best, tp_best,
                                        rsi_y=rsi_y_train_best, rsi_x=rsi_x_train_best, rsi_threshold=rsi_thresh_best)
        tlog_train_rsi   = build_trade_log(df_train, res_train_rsi)

        # ── SAVE EXCEL ─────────────────────────────────────────────────────────
        wb_out          = Workbook()
        pair_type_label = "cointegration" if is_cointegration else "pearson_spearman"
        write_excel(wb_out, A, B, df_test, res_test_base, res_test_rsi,
                    tlog_test_base, tlog_test_rsi, gs_base_df, gs_rsi_df,
                    window, z_atas, z_exit, sl_best, tp_best,
                    rsi_period_best, rsi_thresh_best,
                    rsi_y_series=rsi_y_test_final, rsi_x_series=rsi_x_test_final,
                    tlog_train_base=tlog_train_base, tlog_train_rsi=tlog_train_rsi,
                    pair_type=pair_type_label, direction_info=direction_info)

        output_path = f"BT_RSI/{A}-{B}.xlsx"
        wb_out.save(output_path)
        print(f"  Saved: {output_path}")

    except Exception as e:
        import traceback
        print(f"  Error on {pair}: {e}")
        traceback.print_exc()
        failed_pairs.append((pair, str(e)))

# ── FINAL SUMMARY ──────────────────────────────────────────────────────────────
print(f"\n{'='*65}")
print(f"Completed: {len(all_pairs) - len(failed_pairs)}/{len(all_pairs)} pairs successful")
if failed_pairs:
    print("Failed pairs:")
    for p, reason in failed_pairs:
        print(f"   - {p}: {reason}")

if direction_log:
    print(f"\n{'='*65}")
    print("DIRECTION SELECTION LOG (Pearson/Spearman Pairs)")
    print(f"{'='*65}")
    df_dir_log = pd.DataFrame(direction_log)
    print(df_dir_log.to_string(index=False))
    df_dir_log.to_csv("BT_RSI/direction_selection_log.csv", index=False)
    print(f"\nDirection log saved to: BT_RSI/direction_selection_log.csv")
```
